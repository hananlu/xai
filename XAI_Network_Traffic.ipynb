## 3. Konfigurasi Sampling

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

pd.set_option("display.max_columns", None)

RANDOM_STATE = 42
CHUNK_SIZE = 100_000

FILE_PATH = Path(
    "/content/drive/MyDrive/Machine Learning/"
    "Dataset-Unicauca-Version2-87Atts.csv"
)

TARGET_COLUMN = "ProtocolName"

# Notebook Anda sebelumnya menggunakan Top-10.
# Ubah menjadi 15 jika ingin mengikuti eksperimen sampling sebelumnya.
TOP_K = 10

TARGET_SAMPLE_SIZE = 100_000
```

```python
if not FILE_PATH.exists():
    raise FileNotFoundError(f"File tidak ditemukan: {FILE_PATH}")

print(f"File ditemukan: {FILE_PATH}")
print(f"Ukuran file: {FILE_PATH.stat().st_size / (1024**3):.2f} GB")
```

### 3.1 Distribusi Global dan Pemilihan Kelas

```python
print("Menghitung distribusi global ProtocolName...")

global_counts = pd.Series(dtype="int64")

for chunk_number, chunk in enumerate(
    pd.read_csv(
        FILE_PATH,
        usecols=[TARGET_COLUMN],
        chunksize=CHUNK_SIZE,
        low_memory=False
    ),
    start=1
):
    chunk[TARGET_COLUMN] = (
        chunk[TARGET_COLUMN]
        .astype(str)
        .str.strip()
    )

    chunk_counts = chunk[TARGET_COLUMN].value_counts()

    global_counts = global_counts.add(
        chunk_counts,
        fill_value=0
    )

global_counts = (
    global_counts
    .astype(int)
    .sort_values(ascending=False)
)

print(f"Jumlah kelas unik: {len(global_counts)}")
display(global_counts.head(20).to_frame("Jumlah"))
```

```python
top_apps = global_counts.head(TOP_K).index.tolist()
top_counts = global_counts.loc[top_apps]

print(f"Top-{TOP_K} aplikasi global:")

for rank, (app, count) in enumerate(
    top_counts.items(),
    start=1
):
    print(f"{rank:2d}. {app:<25} {count:,} baris")
```

```python
def calculate_proportional_quota(
    class_counts: pd.Series,
    target_size: int
) -> pd.Series:
    """
    Menghitung kuota sampel proporsional dengan
    total akhir tepat sama dengan target_size.
    """
    total_available = int(class_counts.sum())
    target_size = min(target_size, total_available)

    raw_quota = (
        target_size
        * class_counts
        / total_available
    )

    quota = np.floor(raw_quota).astype(int)

    remaining = target_size - int(quota.sum())

    fractional_part = (
        raw_quota - quota
    ).sort_values(ascending=False)

    for class_name in fractional_part.index[:remaining]:
        quota.loc[class_name] += 1

    return quota.astype(int)


sample_quota = calculate_proportional_quota(
    top_counts,
    TARGET_SAMPLE_SIZE
)

quota_table = pd.DataFrame({
    "Jumlah_Asli": top_counts,
    "Kuota_Sampel": sample_quota,
    "Proporsi_Asli": (
        top_counts / top_counts.sum()
    ),
    "Proporsi_Sampel": (
        sample_quota / sample_quota.sum()
    )
})

display(quota_table)

print(
    "Total kuota sampel:",
    sample_quota.sum()
)
```

### 3.2 Global Stratified Sampling

```python
print("Memulai global stratified sampling...")

rng = np.random.default_rng(RANDOM_STATE)

# Menyimpan kandidat sampel untuk setiap kelas
reservoir = {
    app: pd.DataFrame()
    for app in top_apps
}

for chunk_number, chunk in enumerate(
    pd.read_csv(
        FILE_PATH,
        chunksize=CHUNK_SIZE,
        low_memory=False
    ),
    start=1
):
    chunk[TARGET_COLUMN] = (
        chunk[TARGET_COLUMN]
        .astype(str)
        .str.strip()
    )

    chunk_filtered = chunk[
        chunk[TARGET_COLUMN].isin(top_apps)
    ].copy()

    if chunk_filtered.empty:
        continue

    # Setiap baris mendapat random key independen
    chunk_filtered["_sampling_key"] = rng.random(
        len(chunk_filtered)
    )

    for app in top_apps:
        quota = int(sample_quota.loc[app])

        if quota <= 0:
            continue

        new_rows = chunk_filtered[
            chunk_filtered[TARGET_COLUMN] == app
        ]

        if new_rows.empty:
            continue

        candidate_rows = pd.concat(
            [
                reservoir[app],
                new_rows
            ],
            ignore_index=True
        )

        # Menyimpan random key terbesar sejumlah kuota
        reservoir[app] = (
            candidate_rows
            .nlargest(quota, "_sampling_key")
        )

    if chunk_number % 10 == 0:
        print(f"Chunk selesai: {chunk_number}")
```

```python
df_sample = pd.concat(
    reservoir.values(),
    ignore_index=True
)

df_sample = (
    df_sample
    .drop(columns=["_sampling_key"], errors="ignore")
    .sample(
        frac=1,
        random_state=RANDOM_STATE
    )
    .reset_index(drop=True)
)

print("Ukuran df_sample:")
print(df_sample.shape)

display(df_sample.head())
```

### 3.3 Validasi Distribusi Sampel

```python
sample_counts = (
    df_sample[TARGET_COLUMN]
    .value_counts()
    .reindex(top_apps)
    .fillna(0)
    .astype(int)
)

comparison = pd.DataFrame({
    "Jumlah_Asli": top_counts,
    "Jumlah_Sampel": sample_counts,
    "Proporsi_Asli": (
        top_counts / top_counts.sum()
    ),
    "Proporsi_Sampel": (
        sample_counts / sample_counts.sum()
    )
})

comparison["Selisih_Absolut"] = (
    comparison["Proporsi_Sampel"]
    - comparison["Proporsi_Asli"]
).abs()

display(comparison)
```

```python
assert len(df_sample) == sample_quota.sum()

assert (
    sample_counts.values
    == sample_quota.values
).all()

max_difference = comparison["Selisih_Absolut"].max()

print(f"Jumlah sampel akhir: {len(df_sample):,}")
print(f"Selisih proporsi maksimum: {max_difference:.8f}")
```

```python
# Missing-value handling must be fitted on training data only.
# If there are no missing values, this step leaves the data unchanged.
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="median")

X_train_imputed = pd.DataFrame(
    imputer.fit_transform(X_train_raw),
    columns=X_train_raw.columns,
    index=X_train_raw.index
)

X_valid_imputed = pd.DataFrame(
    imputer.transform(X_valid_raw),
    columns=X_valid_raw.columns,
    index=X_valid_raw.index
)

X_test_imputed = pd.DataFrame(
    imputer.transform(X_test_raw),
    columns=X_test_raw.columns,
    index=X_test_raw.index
)

print("Missing values setelah imputasi:")
print("Train:", int(X_train_imputed.isna().sum().sum()))
print("Valid:", int(X_valid_imputed.isna().sum().sum()))
print("Test :", int(X_test_imputed.isna().sum().sum()))

# Variance filtering berikutnya menggunakan data yang telah diimputasi.
X_train_raw = X_train_imputed
X_valid_raw = X_valid_imputed
X_test_raw = X_test_imputed
```

```python
from sklearn.feature_selection import VarianceThreshold

variance_filter = VarianceThreshold(
    threshold=0.0
)

X_train_var = pd.DataFrame(
    variance_filter.fit_transform(X_train_raw),
    columns=X_train_raw.columns[
        variance_filter.get_support()
    ],
    index=X_train_raw.index
)

X_valid_var = X_valid_raw[
    X_train_var.columns
]

X_test_var = X_test_raw[
    X_train_var.columns
]

X_train = X_train_var.copy()
X_valid = X_valid_var.copy()
X_test = X_test_var.copy()

print("Fitur setelah variance filtering:")
print("X_train:", X_train.shape)
print("X_valid:", X_valid.shape)
print("X_test :", X_test.shape)
```

```python
corr_matrix = X_train_var.corr().abs()

upper_triangle = corr_matrix.where(
    np.triu(
        np.ones(corr_matrix.shape),
        k=1
    ).astype(bool)
)

correlated_columns = [
    column
    for column in upper_triangle.columns
    if any(
        upper_triangle[column] > 0.90
    )
]

selected_features = [
    column
    for column in X_train_var.columns
    if column not in correlated_columns
]

X_train = X_train_var[selected_features]
X_valid = X_valid_var[selected_features]
X_test = X_test_var[selected_features]

print("Fitur awal:", X.shape[1])
print("Fitur dibuang karena korelasi:", len(correlated_columns))
print("Fitur akhir:", X_train.shape[1])
```

## 9. Model Final XGBoost dari Hyperparameter Terpilih

```python
import numpy as np
import xgboost as xgb

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Parameter final hasil optimasi sebelumnya
BEST_ALPHA = 0.50
BEST_CONFIG_ID = 2

number_of_classes = len(
    label_encoder.classes_
)

# Pastikan base_weights sudah tersedia
base_weights = compute_sample_weight(
    class_weight="balanced",
    y=y_train
)

# Bobot final alpha = 0.50
final_sample_weights = (
    base_weights ** BEST_ALPHA
).astype(np.float32)

# Model final langsung
xai_model = xgb.XGBClassifier(
    n_estimators=600,
    max_depth=8,
    min_child_weight=5,
    learning_rate=0.03,
    subsample=0.90,
    colsample_bytree=0.90,

    objective="multi:softprob",
    num_class=number_of_classes,
    eval_metric="mlogloss",
    tree_method="hist",

    reg_alpha=0.1,
    reg_lambda=1.0,

    random_state=RANDOM_STATE,
    n_jobs=-1,
    early_stopping_rounds=40
)

# Training satu model final
xai_model.fit(
    X_train,
    y_train,
    sample_weight=final_sample_weights,
    eval_set=[
        (X_valid, y_valid)
    ],
    verbose=False
)

print("Model final selesai dilatih.")
print("Alpha:", BEST_ALPHA)
print("Config ID:", BEST_CONFIG_ID)
print("Best iteration:", xai_model.best_iteration)
```

### 9.1 Evaluasi Model pada Test Set

```python
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay)


final_test_pred = xai_model.predict(
    X_test
).astype(int)

test_accuracy = accuracy_score(
    y_test,
    final_test_pred
)

test_balanced_accuracy = balanced_accuracy_score(
    y_test,
    final_test_pred
)

test_macro_f1 = f1_score(
    y_test,
    final_test_pred,
    average="macro"
)

print(f"Test Accuracy          : {test_accuracy:.4f}")
print(f"Test Balanced Accuracy : {test_balanced_accuracy:.4f}")
print(f"Test Macro-F1          : {test_macro_f1:.4f}")

print(
    classification_report(
        y_test,
        final_test_pred,
        target_names=label_encoder.classes_,
        digits=4,
        zero_division=0
    )
)
```

```python
def convert_shap_to_3d(
    shap_values,
    n_classes
):
    """
    Output:
    (n_samples, n_features, n_classes)
    """

    if isinstance(shap_values, list):

        # Format lama SHAP:
        # list[class] dengan shape
        # (n_samples, n_features)

        shap_array = np.stack(
            shap_values,
            axis=-1
        )

    else:

        shap_array = np.asarray(
            shap_values
        )

        # Format multiclass modern
        if shap_array.ndim == 3:
            pass

        # Format single-output
        elif shap_array.ndim == 2:
            shap_array = shap_array[:, :, np.newaxis]

        else:
            raise ValueError(
                f"Shape SHAP tidak dikenali: "
                f"{shap_array.shape}"
            )

    return shap_array


shap_array = convert_shap_to_3d(
    shap_raw,
    n_classes=len(label_encoder.classes_)
)

print("Shape SHAP final:", shap_array.shape)
```

### 11.4 Perbandingan TreeSHAP untuk Kelas Aktual dan Prediksi

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

def plot_local_shap_comparison(
    sample_position,
    shap_array,
    X_xai,
    y_xai,
    xai_pred,
    label_encoder,
    top_n=15
):

    true_class_id = int(y_xai[sample_position])
    predicted_class_id = int(xai_pred[sample_position])

    true_class_name = label_encoder.inverse_transform(
        [true_class_id]
    )[0]

    predicted_class_name = label_encoder.inverse_transform(
        [predicted_class_id]
    )[0]

    print("Actual class   :", true_class_name)
    print("Predicted class:", predicted_class_name)

    # Penjelasan untuk kelas aktual
    true_values = shap_array[
        sample_position,
        :,
        true_class_id
    ]

    true_df = pd.DataFrame({
        "feature": X_xai.columns,
        "shap_value": true_values
    })

    true_df["abs_shap"] = (
        true_df["shap_value"].abs()
    )

    true_df = (
        true_df
        .sort_values(
            "abs_shap",
            ascending=False
        )
        .head(top_n)
        .sort_values(
            "shap_value",
            ascending=True
        )
    )

    # Penjelasan untuk kelas prediksi
    predicted_values = shap_array[
        sample_position,
        :,
        predicted_class_id
    ]

    predicted_df = pd.DataFrame({
        "feature": X_xai.columns,
        "shap_value": predicted_values
    })

    predicted_df["abs_shap"] = (
        predicted_df["shap_value"].abs()
    )

    predicted_df = (
        predicted_df
        .sort_values(
            "abs_shap",
            ascending=False
        )
        .head(top_n)
        .sort_values(
            "shap_value",
            ascending=True
        )
    )

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(16, 7),
        sharey=False
    )

    # Kelas aktual
    colors_true = [
        "steelblue" if value > 0 else "red"
        for value in true_df["shap_value"]
    ]

    axes[0].barh(
        true_df["feature"],
        true_df["shap_value"],
        color=colors_true
    )

    axes[0].axvline(
        0,
        color="black",
        linewidth=0.8
    )

    axes[0].set_title(
        f"Explanation for Actual Class\n"
        f"{true_class_name}"
    )

    axes[0].set_xlabel("SHAP Value")
    axes[0].set_ylabel("Feature")

    # Kelas prediksi
    colors_pred = [
        "steelblue" if value > 0 else "red"
        for value in predicted_df["shap_value"]
    ]

    axes[1].barh(
        predicted_df["feature"],
        predicted_df["shap_value"],
        color=colors_pred
    )

    axes[1].axvline(
        0,
        color="black",
        linewidth=0.8
    )

    axes[1].set_title(
        f"Explanation for Predicted Class\n"
        f"{predicted_class_name}"
    )

    axes[1].set_xlabel("SHAP Value")
    axes[1].set_ylabel("Feature")

    fig.suptitle(
        f"TreeSHAP Local Comparison\n"
        f"Actual: {true_class_name} | "
        f"Predicted: {predicted_class_name}",
        fontsize=14
    )

    plt.tight_layout()
    plt.show()

    return true_df, predicted_df
```

```python
import re

def extract_base_feature(expression, feature_names):
    """
    Mengambil nama fitur asli dari ekspresi LIME.
    """

    # Prioritaskan nama fitur yang lebih panjang
    sorted_features = sorted(
        feature_names,
        key=len,
        reverse=True
    )

    for feature in sorted_features:
        if feature in expression:
            return feature

    return expression
```

## 14. Per-sample Agreement TreeSHAP dan LIME

```python
from scipy.stats import kendalltau

def top_shap_features_for_sample(sample_position, top_n=10):
    predicted_class_id = int(xai_pred[sample_position])

    local_values = shap_array[
        sample_position,
        :,
        predicted_class_id
    ]

    local_df = pd.DataFrame({
        "feature": X_xai.columns,
        "value": local_values
    })

    return (
        local_df.assign(abs_value=local_df["value"].abs())
        .sort_values("abs_value", ascending=False)
        .head(top_n)["feature"].tolist()
    )


def top_lime_features_for_sample(sample_position, top_n=10):
    predicted_class_id = int(xai_pred[sample_position])
    sample = X_xai.iloc[sample_position].to_numpy()

    explanation = lime_explainer.explain_instance(
        data_row=sample,
        predict_fn=xai_model.predict_proba,
        num_features=top_n,
        labels=(predicted_class_id,)
    )

    lime_list = explanation.as_list(
        label=predicted_class_id
    )

    features = [
        extract_base_feature(
            expression,
            list(X_xai.columns)
        )
        for expression, _ in lime_list
    ]

    return list(dict.fromkeys(features))[:top_n]


def jaccard_similarity(features_a, features_b):
    set_a = set(features_a)
    set_b = set(features_b)
    union = set_a.union(set_b)
    return (
        len(set_a.intersection(set_b)) / len(union)
        if union else 0.0
    )


agreement_records = []

for position in range(len(X_xai)):
    shap_features = top_shap_features_for_sample(position)
    lime_features = top_lime_features_for_sample(position)

    agreement_records.append({
        "sample_position": position,
        "true_class": label_encoder.inverse_transform(
            [int(y_xai[position])]
        )[0],
        "predicted_class": label_encoder.inverse_transform(
            [int(xai_pred[position])]
        )[0],
        "correct": bool(y_xai[position] == xai_pred[position]),
        "jaccard_top10": jaccard_similarity(
            shap_features,
            lime_features
        )
    })

agreement_df = pd.DataFrame(agreement_records)

display(agreement_df.head())
print("Mean Jaccard:", agreement_df["jaccard_top10"].mean())
print("Median Jaccard:", agreement_df["jaccard_top10"].median())

display(
    agreement_df
    .groupby("correct")["jaccard_top10"]
    .agg(["mean", "median", "std", "count"])
)

agreement_df.to_csv(
    "treeshap_lime_per_sample_agreement.csv",
    index=False
)
```

Visualisasi ini menggunakan sampel, kelas prediksi, jumlah fitur, dan arah warna yang sama dengan TreeSHAP. Nilai SHAP dan bobot LIME tidak dibandingkan secara langsung; perbandingan difokuskan pada fitur, ranking, dan arah kontribusi.


```python
def get_lime_local_dataframe(
    sample_position,
    top_n=15,
    explainer=None
):
    if explainer is None:
        explainer = lime_explainer

    predicted_class_id = int(
        xai_pred[sample_position]
    )

    sample = X_xai.iloc[
        sample_position
    ].to_numpy()

    explanation = explainer.explain_instance(
        data_row=sample,
        predict_fn=xai_model.predict_proba,
        num_features=top_n,
        labels=(predicted_class_id,)
    )

    lime_list = explanation.as_list(
        label=predicted_class_id
    )

    local_df = pd.DataFrame(
        lime_list,
        columns=[
            "feature_expression",
            "lime_weight"
        ]
    )

    local_df["feature"] = local_df[
        "feature_expression"
    ].apply(
        lambda expression: extract_base_feature(
            expression,
            list(X_xai.columns)
        )
    )

    # Satu fitur dapat muncul dalam lebih dari satu interval.
    local_df = (
        local_df
        .groupby("feature", as_index=False)
        .agg(
            lime_weight=("lime_weight", "sum")
        )
    )

    local_df["abs_lime_weight"] = (
        local_df["lime_weight"].abs()
    )

    return explanation, local_df


def plot_lime_local(
    sample_position,
    top_n=15,
    output_file="figure_lime_local.png"
):
    true_class_id = int(y_xai[sample_position])
    predicted_class_id = int(xai_pred[sample_position])

    true_name = label_encoder.inverse_transform(
        [true_class_id]
    )[0]

    predicted_name = label_encoder.inverse_transform(
        [predicted_class_id]
    )[0]

    explanation, local_df = (
        get_lime_local_dataframe(
            sample_position,
            top_n=top_n
        )
    )

    plot_df = (
        local_df
        .sort_values(
            "abs_lime_weight",
            ascending=False
        )
        .head(top_n)
        .sort_values(
            "lime_weight",
            ascending=True
        )
    )

    colors = [
        "steelblue" if value > 0 else "red"
        for value in plot_df["lime_weight"]
    ]

    fig, ax = plt.subplots(figsize=(9, 6))

    ax.barh(
        plot_df["feature"],
        plot_df["lime_weight"],
        color=colors
    )

    ax.axvline(
        0,
        color="black",
        linewidth=0.8
    )

    ax.set_xlabel("LIME Weight")
    ax.set_ylabel("Feature")
    ax.set_title(
        "LIME Local Explanation\n"
        f"Actual: {true_name} | Predicted: {predicted_name}"
    )

    plt.tight_layout()
    plt.savefig(
        output_file,
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()

    return explanation, local_df


lime_example_exp, lime_example_df = plot_lime_local(
    sample_position=incorrect_indices[0],
    top_n=15,
    output_file="figure_lime_local.png"
)
```

## 18. Agreement TreeSHAP–LIME per Kelas dan Kendall Correlation


```python
from scipy.stats import kendalltau


def kendall_topk(features_a, features_b):
    common = [
        feature for feature in features_a
        if feature in features_b
    ]

    if len(common) < 2:
        return np.nan

    ranks_a = [
        features_a.index(feature)
        for feature in common
    ]

    ranks_b = [
        features_b.index(feature)
        for feature in common
    ]

    tau, _ = kendalltau(ranks_a, ranks_b)
    return tau


# Tambahkan Kendall correlation ke agreement per sampel.
agreement_records_kendall = []

for position in range(len(X_xai)):
    shap_features = top_shap_features_for_sample(position, top_n=10)
    lime_features = top_lime_features_for_sample(position, top_n=10)

    agreement_records_kendall.append({
        "sample_position": position,
        "true_class": label_encoder.inverse_transform(
            [int(y_xai[position])]
        )[0],
        "predicted_class": label_encoder.inverse_transform(
            [int(xai_pred[position])]
        )[0],
        "correct": bool(y_xai[position] == xai_pred[position]),
        "jaccard_top10": jaccard_similarity(
            shap_features,
            lime_features
        ),
        "kendall_tau": kendall_topk(
            shap_features,
            lime_features
        )
    })

agreement_kendall_df = pd.DataFrame(
    agreement_records_kendall
)

display(agreement_kendall_df.head())

print(
    "Mean Jaccard:",
    agreement_kendall_df["jaccard_top10"].mean()
)

print(
    "Mean Kendall tau:",
    agreement_kendall_df["kendall_tau"].mean()
)

display(
    agreement_kendall_df
    .groupby("true_class")[
        ["jaccard_top10", "kendall_tau"]
    ]
    .agg(["mean", "median", "std", "count"])
)

agreement_kendall_df.to_csv(
    "treeshap_lime_agreement_kendall.csv",
    index=False
)
```

## 19. Stability LIME

LIME bersifat stochastic. Stability diukur dengan menjalankan LIME beberapa kali pada sampel yang sama dan menghitung Jaccard antar-rerun. Untuk menjaga waktu eksekusi, digunakan lima sampel per kelas.


```python
STABILITY_RUNS = 5
STABILITY_SAMPLES_PER_CLASS = 5

stability_positions = []

for class_id in np.unique(y_xai):
    positions = np.where(y_xai == class_id)[0]
    stability_positions.extend(
        positions[:STABILITY_SAMPLES_PER_CLASS]
    )

stability_records = []

for position in stability_positions:
    repeated_top_features = []

    for run_id in range(STABILITY_RUNS):
        stability_explainer = LimeTabularExplainer(
            training_data=X_train.sample(
                n=min(10000, len(X_train)),
                random_state=RANDOM_STATE
            ).to_numpy(),
            feature_names=list(X_train.columns),
            class_names=list(label_encoder.classes_),
            mode="classification",
            discretize_continuous=True,
            random_state=RANDOM_STATE + run_id
        )

        _, run_df = get_lime_local_dataframe(
            position,
            top_n=10,
            explainer=stability_explainer
        )

        repeated_top_features.append(
            run_df
            .sort_values(
                "abs_lime_weight",
                ascending=False
            )
            .head(10)["feature"].tolist()
        )

    pair_scores = []

    for i in range(len(repeated_top_features)):
        for j in range(i + 1, len(repeated_top_features)):
            pair_scores.append(
                jaccard_similarity(
                    repeated_top_features[i],
                    repeated_top_features[j]
                )
            )

    stability_records.append({
        "sample_position": position,
        "true_class": label_encoder.inverse_transform(
            [int(y_xai[position])]
        )[0],
        "mean_pairwise_jaccard": np.mean(pair_scores),
        "std_pairwise_jaccard": np.std(pair_scores)
    })

stability_df = pd.DataFrame(stability_records)
display(stability_df.head())

print(
    "Mean LIME stability:",
    stability_df["mean_pairwise_jaccard"].mean()
)

stability_df.to_csv(
    "lime_stability_results.csv",
    index=False
)
```

## 20. Faithfulness melalui Feature Removal

Fitur penting dihapus dengan menggantinya menggunakan median training data. Penurunan probabilitas kelas prediksi digunakan sebagai ukuran faithfulness. Semakin besar penurunan, semakin besar pengaruh fitur yang dianggap penting terhadap prediksi model.


```python
def get_top_pfi_features(top_n=10):
    return permutation_df.head(top_n)["feature"].tolist()


def probability_drop_for_method(
    method,
    top_k=5
):
    baseline_proba = xai_model.predict_proba(X_xai)
    predicted_ids = xai_pred.astype(int)
    rows = np.arange(len(X_xai))

    baseline_selected = baseline_proba[
        rows,
        predicted_ids
    ]

    X_modified = X_xai.copy()
    train_medians = X_train.median(numeric_only=True)

    for position in range(len(X_xai)):

        if method == "TreeSHAP":
            features = top_shap_features_for_sample(
                position,
                top_n=top_k
            )

        elif method == "LIME":
            features = top_lime_features_for_sample(
                position,
                top_n=top_k
            )

        elif method == "Permutation":
            features = get_top_pfi_features(
                top_n=top_k
            )

        else:
            raise ValueError(
                f"Metode tidak dikenali: {method}"
            )

        for feature in features:
            if feature in X_modified.columns:
                X_modified.loc[
                    X_modified.index[position],
                    feature
                ] = train_medians[feature]

    modified_proba = xai_model.predict_proba(
        X_modified
    )

    modified_selected = modified_proba[
        rows,
        predicted_ids
    ]

    return np.mean(
        baseline_selected - modified_selected
    )


faithfulness_records = []

for method in [
    "TreeSHAP",
    "LIME",
    "Permutation"
]:
    for top_k in [1, 3, 5, 10]:
        faithfulness_records.append({
            "method": method,
            "top_k": top_k,
            "mean_probability_drop": probability_drop_for_method(
                method,
                top_k=top_k
            )
        })

faithfulness_df = pd.DataFrame(
    faithfulness_records
)

display(faithfulness_df)

faithfulness_df.to_csv(
    "xai_faithfulness_results.csv",
    index=False
)
```

## 24. Confidence Interval untuk Evaluasi Model

Confidence interval digunakan agar performa model tidak hanya dilaporkan sebagai satu angka. Bootstrap dilakukan secara stratified sehingga setiap kelas tetap terwakili pada setiap pengulangan.


```python
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score
)


def stratified_bootstrap_indices(
    y_true,
    rng
):
    y_true = np.asarray(y_true)
    selected = []

    for class_id in np.unique(y_true):
        class_indices = np.where(
            y_true == class_id
        )[0]

        selected.extend(
            rng.choice(
                class_indices,
                size=len(class_indices),
                replace=True
            ).tolist()
        )

    return np.asarray(selected)


def bootstrap_metric_ci(
    y_true,
    y_pred,
    metric_function,
    n_bootstrap=1000,
    random_state=42
):
    rng = np.random.default_rng(random_state)
    bootstrap_values = []

    for _ in range(n_bootstrap):
        indices = stratified_bootstrap_indices(
            y_true,
            rng
        )

        bootstrap_values.append(
            metric_function(
                np.asarray(y_true)[indices],
                np.asarray(y_pred)[indices]
            )
        )

    values = np.asarray(bootstrap_values)

    return {
        "mean": np.mean(values),
        "lower_95": np.percentile(values, 2.5),
        "upper_95": np.percentile(values, 97.5),
        "std": np.std(values)
    }


metrics_for_ci = {
    "accuracy": accuracy_score,
    "balanced_accuracy": balanced_accuracy_score,
    "macro_f1": lambda y, p: f1_score(
        y,
        p,
        average="macro",
        zero_division=0
    ),
    "weighted_f1": lambda y, p: f1_score(
        y,
        p,
        average="weighted",
        zero_division=0
    )
}

ci_records = []

for metric_name, metric_function in metrics_for_ci.items():
    ci = bootstrap_metric_ci(
        y_test,
        final_test_pred,
        metric_function,
        n_bootstrap=1000,
        random_state=RANDOM_STATE
    )

    ci_records.append({
        "metric": metric_name,
        **ci
    })

metrics_ci_df = pd.DataFrame(ci_records)
display(metrics_ci_df)

metrics_ci_df.to_csv(
    "classification_metrics_bootstrap_ci.csv",
    index=False
)
```
