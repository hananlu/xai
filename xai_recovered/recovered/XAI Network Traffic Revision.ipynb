```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

pd.set_option("display.max_columns", None)

RANDOM_STATE = 42
CHUNK_SIZE = 100_000

FILE_PATH = Path(
    "/content/drive/MyDrive/Machine Learning/"
    "Dataset-Unicauca-Version2-87Atts.csv"
)

TARGET_COLUMN = "ProtocolName"

# Notebook Anda sebelumnya menggunakan Top-10.
# Ubah menjadi 15 jika ingin mengikuti eksperimen sampling sebelumnya.
TOP_K = 10

TARGET_SAMPLE_SIZE = 100_000
```

```python
if not FILE_PATH.exists():
    raise FileNotFoundError(f"File tidak ditemukan: {FILE_PATH}")

print(f"File ditemukan: {FILE_PATH}")
print(f"Ukuran file: {FILE_PATH.stat().st_size / (1024**3):.2f} GB")
```

```python
if TARGET_COLUMN not in header.columns:
    raise KeyError(
        f"Kolom target {TARGET_COLUMN!r} tidak ditemukan."
    )
```

```python
print("Menghitung distribusi global ProtocolName...")

global_counts = pd.Series(dtype="int64")

for chunk_number, chunk in enumerate(
    pd.read_csv(
        FILE_PATH,
        usecols=[TARGET_COLUMN],
        chunksize=CHUNK_SIZE,
        low_memory=False
    ),
    start=1
):
    chunk[TARGET_COLUMN] = (
        chunk[TARGET_COLUMN]
        .astype(str)
        .str.strip()
    )

    chunk_counts = chunk[TARGET_COLUMN].value_counts()

    global_counts = global_counts.add(
        chunk_counts,
        fill_value=0
    )

global_counts = (
    global_counts
    .astype(int)
    .sort_values(ascending=False)
)

print(f"Jumlah kelas unik: {len(global_counts)}")
display(global_counts.head(20).to_frame("Jumlah"))
```

```python
top_apps = global_counts.head(TOP_K).index.tolist()
top_counts = global_counts.loc[top_apps]

print(f"Top-{TOP_K} aplikasi global:")

for rank, (app, count) in enumerate(
    top_counts.items(),
    start=1
):
    print(f"{rank:2d}. {app:<25} {count:,} baris")
```

```python
def calculate_proportional_quota(
    class_counts: pd.Series,
    target_size: int
) -> pd.Series:
    """
    Menghitung kuota sampel proporsional dengan
    total akhir tepat sama dengan target_size.
    """
    total_available = int(class_counts.sum())
    target_size = min(target_size, total_available)

    raw_quota = (
        target_size
        * class_counts
        / total_available
    )

    quota = np.floor(raw_quota).astype(int)

    remaining = target_size - int(quota.sum())

    fractional_part = (
        raw_quota - quota
    ).sort_values(ascending=False)

    for class_name in fractional_part.index[:remaining]:
        quota.loc[class_name] += 1

    return quota.astype(int)


sample_quota = calculate_proportional_quota(
    top_counts,
    TARGET_SAMPLE_SIZE
)

quota_table = pd.DataFrame({
    "Jumlah_Asli": top_counts,
    "Kuota_Sampel": sample_quota,
    "Proporsi_Asli": (
        top_counts / top_counts.sum()
    ),
    "Proporsi_Sampel": (
        sample_quota / sample_quota.sum()
    )
})

display(quota_table)

print(
    "Total kuota sampel:",
    sample_quota.sum()
)
```

```python
print("Memulai global stratified sampling...")

rng = np.random.default_rng(RANDOM_STATE)

# Menyimpan kandidat sampel untuk setiap kelas
reservoir = {
    app: pd.DataFrame()
    for app in top_apps
}

for chunk_number, chunk in enumerate(
    pd.read_csv(
        FILE_PATH,
        chunksize=CHUNK_SIZE,
        low_memory=False
    ),
    start=1
):
    chunk[TARGET_COLUMN] = (
        chunk[TARGET_COLUMN]
        .astype(str)
        .str.strip()
    )

    chunk_filtered = chunk[
        chunk[TARGET_COLUMN].isin(top_apps)
    ].copy()

    if chunk_filtered.empty:
        continue

    # Setiap baris mendapat random key independen
    chunk_filtered["_sampling_key"] = rng.random(
        len(chunk_filtered)
    )

    for app in top_apps:
        quota = int(sample_quota.loc[app])

        if quota <= 0:
            continue

        new_rows = chunk_filtered[
            chunk_filtered[TARGET_COLUMN] == app
        ]

        if new_rows.empty:
            continue

        candidate_rows = pd.concat(
            [
                reservoir[app],
                new_rows
            ],
            ignore_index=True
        )

        # Menyimpan random key terbesar sejumlah kuota
        reservoir[app] = (
            candidate_rows
            .nlargest(quota, "_sampling_key")
        )

    if chunk_number % 10 == 0:
        print(f"Chunk selesai: {chunk_number}")
```

```python
df_sample = pd.concat(
    reservoir.values(),
    ignore_index=True
)

df_sample = (
    df_sample
    .drop(columns=["_sampling_key"], errors="ignore")
    .sample(
        frac=1,
        random_state=RANDOM_STATE
    )
    .reset_index(drop=True)
)

print("Ukuran df_sample:")
print(df_sample.shape)

display(df_sample.head())
```

```python
sample_counts = (
    df_sample[TARGET_COLUMN]
    .value_counts()
    .reindex(top_apps)
    .fillna(0)
    .astype(int)
)

comparison = pd.DataFrame({
    "Jumlah_Asli": top_counts,
    "Jumlah_Sampel": sample_counts,
    "Proporsi_Asli": (
        top_counts / top_counts.sum()
    ),
    "Proporsi_Sampel": (
        sample_counts / sample_counts.sum()
    )
})

comparison["Selisih_Absolut"] = (
    comparison["Proporsi_Sampel"]
    - comparison["Proporsi_Asli"]
).abs()

display(comparison)
```

```python
assert len(df_sample) == sample_quota.sum()

assert (
    sample_counts.values
    == sample_quota.values
).all()

max_difference = comparison["Selisih_Absolut"].max()

print(f"Jumlah sampel akhir: {len(df_sample):,}")
print(f"Selisih proporsi maksimum: {max_difference:.8f}")
```

```python
import numpy as np
import xgboost as xgb

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Parameter final hasil optimasi sebelumnya
BEST_ALPHA = 0.50
BEST_CONFIG_ID = 2

number_of_classes = len(
    label_encoder.classes_
)

# Pastikan base_weights sudah tersedia
base_weights = compute_sample_weight(
    class_weight="balanced",
    y=y_train
)

# Bobot final alpha = 0.50
final_sample_weights = (
    base_weights ** BEST_ALPHA
).astype(np.float32)

# Model final langsung
xai_model = xgb.XGBClassifier(
    n_estimators=600,
    max_depth=8,
    min_child_weight=5,
    learning_rate=0.03,
    subsample=0.90,
    colsample_bytree=0.90,

    objective="multi:softprob",
    num_class=number_of_classes,
    eval_metric="mlogloss",
    tree_method="hist",

    reg_alpha=0.1,
    reg_lambda=1.0,

    random_state=RANDOM_STATE,
    n_jobs=-1,
    early_stopping_rounds=40
)

# Training satu model final
xai_model.fit(
    X_train,
    y_train,
    sample_weight=final_sample_weights,
    eval_set=[
        (X_valid, y_valid)
    ],
    verbose=False
)

print("Model final selesai dilatih.")
print("Alpha:", BEST_ALPHA)
print("Config ID:", BEST_CONFIG_ID)
print("Best iteration:", xai_model.best_iteration)
```

```python
def convert_shap_to_3d(
    shap_values,
    n_classes
):
    """
    Output:
    (n_samples, n_features, n_classes)
    """

    if isinstance(shap_values, list):

        # Format lama SHAP:
        # list[class] dengan shape
        # (n_samples, n_features)

        shap_array = np.stack(
            shap_values,
            axis=-1
        )

    else:

        shap_array = np.asarray(
            shap_values
        )

        # Format multiclass modern
        if shap_array.ndim == 3:
            pass

        # Format single-output
        elif shap_array.ndim == 2:
            shap_array = shap_array[:, :, np.newaxis]

        else:
            raise ValueError(
                f"Shape SHAP tidak dikenali: "
                f"{shap_array.shape}"
            )

    return shap_array


shap_array = convert_shap_to_3d(
    shap_raw,
    n_classes=len(label_encoder.classes_)
)

print("Shape SHAP final:", shap_array.shape)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

def plot_local_shap_comparison(
    sample_position,
    shap_array,
    X_xai,
    y_xai,
    xai_pred,
    label_encoder,
    top_n=15
):

    true_class_id = int(y_xai[sample_position])
    predicted_class_id = int(xai_pred[sample_position])

    true_class_name = label_encoder.inverse_transform(
        [true_class_id]
    )[0]

    predicted_class_name = label_encoder.inverse_transform(
        [predicted_class_id]
    )[0]

    print("Actual class   :", true_class_name)
    print("Predicted class:", predicted_class_name)

    # Penjelasan untuk kelas aktual
    true_values = shap_array[
        sample_position,
        :,
        true_class_id
    ]

    true_df = pd.DataFrame({
        "feature": X_xai.columns,
        "shap_value": true_values
    })

    true_df["abs_shap"] = (
        true_df["shap_value"].abs()
    )

    true_df = (
        true_df
        .sort_values(
            "abs_shap",
            ascending=False
        )
        .head(top_n)
        .sort_values(
            "shap_value",
            ascending=True
        )
    )

    # Penjelasan untuk kelas prediksi
    predicted_values = shap_array[
        sample_position,
        :,
        predicted_class_id
    ]

    predicted_df = pd.DataFrame({
        "feature": X_xai.columns,
        "shap_value": predicted_values
    })

    predicted_df["abs_shap"] = (
        predicted_df["shap_value"].abs()
    )

    predicted_df = (
        predicted_df
        .sort_values(
            "abs_shap",
            ascending=False
        )
        .head(top_n)
        .sort_values(
            "shap_value",
            ascending=True
        )
    )

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(16, 7),
        sharey=False
    )

    # Kelas aktual
    colors_true = [
        "steelblue" if value > 0 else "red"
        for value in true_df["shap_value"]
    ]

    axes[0].barh(
        true_df["feature"],
        true_df["shap_value"],
        color=colors_true
    )

    axes[0].axvline(
        0,
        color="black",
        linewidth=0.8
    )

    axes[0].set_title(
        f"Explanation for Actual Class\n"
        f"{true_class_name}"
    )

    axes[0].set_xlabel("SHAP Value")
    axes[0].set_ylabel("Feature")

    # Kelas prediksi
    colors_pred = [
        "steelblue" if value > 0 else "red"
        for value in predicted_df["shap_value"]
    ]

    axes[1].barh(
        predicted_df["feature"],
        predicted_df["shap_value"],
        color=colors_pred
    )

    axes[1].axvline(
        0,
        color="black",
        linewidth=0.8
    )

    axes[1].set_title(
        f"Explanation for Predicted Class\n"
        f"{predicted_class_name}"
    )

    axes[1].set_xlabel("SHAP Value")
    axes[1].set_ylabel("Feature")

    fig.suptitle(
        f"TreeSHAP Local Comparison\n"
        f"Actual: {true_class_name} | "
        f"Predicted: {predicted_class_name}",
        fontsize=14
    )

    plt.tight_layout()
    plt.show()

    return true_df, predicted_df
```

```python
import re

def extract_base_feature(expression, feature_names):
    """
    Mengambil nama fitur asli dari ekspresi LIME.
    """

    # Prioritaskan nama fitur yang lebih panjang
    sorted_features = sorted(
        feature_names,
        key=len,
        reverse=True
    )

    for feature in sorted_features:
        if feature in expression:
            return feature

    return expression
```

```python
lime_df["feature"] = lime_df[
    "feature_expression"
].apply(
    lambda expression: extract_base_feature(
        expression,
        list(X_xai.columns)
    )
)

display(
    lime_df[
        [
            "feature_expression",
            "feature",
            "weight"
        ]
    ].head()
)
```

```python
lime_global_df = (
    lime_df
    .groupby("feature")
    .agg(
        mean_abs_lime=("abs_weight", "mean"),
        mean_lime=("weight", "mean"),
        std_lime=("weight", "std")
    )
    .reset_index()
    .sort_values(
        "mean_abs_lime",
        ascending=False
    )
)

display(
    lime_global_df.head(20)
)
```

```python
top_k = 10

top_shap_features = (
    shap_global_df
    .head(top_k)["feature"]
    .tolist()
)

top_lime_features = (
    lime_global_df
    .head(top_k)["feature"]
    .tolist()
)

def jaccard_similarity(features_a, features_b):
    set_a = set(features_a)
    set_b = set(features_b)

    union = set_a.union(set_b)

    if len(union) == 0:
        return 0.0

    return len(
        set_a.intersection(set_b)
    ) / len(union)

jaccard_score = jaccard_similarity(
    top_shap_features,
    top_lime_features
)

print("Top TreeSHAP:")
print(top_shap_features)

print("\nTop LIME:")
print(top_lime_features)

print(
    "\nJaccard similarity Top-10:",
    round(jaccard_score, 4)
)
```
