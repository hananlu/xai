```python
file_path = '/content/drive/MyDrive/Machine Learning/Dataset-Unicauca-Version2-87Atts.csv'
print("[STEP 1] Mengidentifikasi aplikasi terpopuler dari dataset...")
# Baca kolom ProtocolName saja dari 200.000 baris pertama untuk identifikasi cepat
df_proto_preview = pd.read_csv(file_path, usecols=['ProtocolName'], nrows=200000)
top_apps = df_proto_preview['ProtocolName'].value_counts().head(10).index.tolist()

print("\n10 Aplikasi Terbanyak Teridentifikasi:")
for idx, app in enumerate(top_apps, 1):
    print(f"{idx}. {app}")
```

```python
print("\n[STEP 2] Mengambil Stratified Sample (~100.000 baris)...")
chunk_size = 100000
sampled_chunks = []
target_sample = 100000

# Baca per-chunk untuk menjaga kestabilan RAM Colab
for chunk in pd.read_csv(file_path, chunksize=chunk_size, low_memory=False):
    # Standarisasi penamaan kolom
    chunk.columns = chunk.columns.str.strip().str.replace(' ', '_').str.replace('.', '_')

    # Filter hanya baris yang termasuk dalam Top-10 aplikasi
    chunk_filtered = chunk[chunk['ProtocolName'].isin(top_apps)]

    if len(chunk_filtered) > 0:
        # Sampling proporsional 4% dari setiap potongan file
        sample = chunk_filtered.sample(frac=0.04, random_state=42)
        sampled_chunks.append(sample)

    if sum(len(c) for c in sampled_chunks) >= target_sample:
        break

df_sample = pd.concat(sampled_chunks, ignore_index=True)
print(f"Sampel berhasil dibuat: {df_sample.shape[0]} baris dan {df_sample.shape[1]} kolom.")
print("\nDistribusi Sampel Kelas Target (ProtocolName):")
print(df_sample['ProtocolName'].value_counts())
```

```python
print("\n[STEP 3] Pemeriksaan Missing Values dan Infinite Values...")

# 1. Hapus pengenal statis (Anti-Leakage) dan kolom zero-variance teridentifikasi
cols_to_remove = [
    'Flow_ID', 'Source_IP', 'Source_Port',
    'Destination_IP', 'Destination_Port', 'Timestamp',
    'Label', 'L7Protocol', 'Fwd_Header_Length_1',
    'Bwd_PSH_Flags', 'Fwd_URG_Flags', 'Bwd_URG_Flags', 'CWE_Flag_Count',
    'Fwd_Avg_Bytes_Bulk', 'Fwd_Avg_Packets_Bulk', 'Fwd_Avg_Bulk_Rate',
    'Bwd_Avg_Bytes_Bulk', 'Bwd_Avg_Packets_Bulk', 'Bwd_Avg_Bulk_Rate'
]
df_clean = df_sample.drop(columns=[c for c in cols_to_remove if c in df_sample.columns], errors='ignore')
```

```python
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
```

```python
X = df_clean.drop('ProtocolName', axis=1)
y = df_clean['ProtocolName']
```

```python
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y)
```

```python
# Tampilkan kamus kelas
class_mapping = dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_)))
print("\nPemetaan Kelas Aplikasi (Label Encoding):")
for name, idx in list(class_mapping.items())[:10]:
    print(f"  {idx} -> {name}")
```

```python
print("\nMenyaring fitur multikolinearitas tinggi...")
corr_matrix = X.corr().abs()
upper_tri = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

# Cari kolom dengan korelasi di atas 0.90
multicollinear_cols = [col for col in upper_tri.columns if any(upper_tri[col] > 0.90)]
X_filtered = X.drop(columns=multicollinear_cols)

print(f"Jumlah fitur awal        : {X.shape[1]}")
print(f"Fitur redundan dibuang   : {len(multicollinear_cols)}")
print(f"Fitur aktif dipertahankan: {X_filtered.shape[1]}")
```

```python
print("\nMelakukan partisi data terstratifikasi (70% Train, 30% Test)...")
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X_filtered, y,
    test_size=0.30,
    random_state=42,
    stratify=y
)
```

```python
scaler = StandardScaler()

# Fit hanya pada data latih, lalu transformasi ke data latih dan data uji
X_train = pd.DataFrame(
    scaler.fit_transform(X_train_raw),
    columns=X_filtered.columns
)
X_test = pd.DataFrame(
    scaler.transform(X_test_raw),
    columns=X_filtered.columns
)

print("\n--- RINGKASAN DATA SIAP LATIH ---")
print(f"Dimensi X_train : {X_train.shape}")
print(f"Dimensi X_test  : {X_test.shape}")
print(f"Dimensi y_train : {y_train.shape}")
print(f"Dimensi y_test  : {y_test.shape}")
print("Pra-pemrosesan selesai 100%. Data siap dimasukkan ke model XGBoost!")
```

```python
import xgboost as xgb
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.metrics import classification_report, accuracy_score, f1_score
import time
```

```python
print("--- TAHAP 1: PELATIHAN MODEL XGBOOST ---")

# 1. Menentukan parameter model XGBoost
# n_estimators=100 dan max_depth=5 memberikan kompromi optimal antara performa dan waktu latih
num_classes = len(np.unique(y_train))
objective = 'multi:softprob' if num_classes > 2 else 'binary:logistic'

xgb_model = xgb.XGBClassifier(
    n_estimators=100,
    max_depth=5,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    objective=objective,
    random_state=42,
    eval_metric='mlogloss' if num_classes > 2 else 'logloss',
    n_jobs=-1
)
```

```python
# 2. Melatih model pada data latih
t_start = time.time()
print("Sedang melatih model XGBoost pada 69.856 sampel data...")
xgb_model.fit(X_train, y_train)
t_elapsed = time.time() - t_start
print(f"Pelatihan selesai dalam {t_elapsed:.2f} detik!")
```

```python
# 3. Prediksi pada data uji
print("\nMengevaluasi model pada 29.939 data uji...")
y_pred = xgb_model.predict(X_test)
```

```python
# 4. Tampilkan metrik kinerja klasifikasi
acc = accuracy_score(y_test, y_pred)
macro_f1 = f1_score(y_test, y_pred, average='macro')

print("\n" + "="*60)
print(f"Akurasi Keseluruhan (Accuracy) : {acc * 100:.2f}%")
print(f"Macro F1-Score                 : {macro_f1:.4f}")
print("="*60)

# Tampilkan classification report menggunakan nama asli aplikasi (ProtocolName)
target_names = [str(cls) for cls in label_encoder.classes_]
print("\nLaporan Klasifikasi Detail per Aplikasi:")
print(classification_report(y_test, y_pred, target_names=target_names, digits=4))
```

```python
print("--- MELATIH ULANG XGBOOST DENGAN SAMPLE WEIGHT BALANCED ---")

# 1. Hitung bobot sampel seimbang untuk mengatasi ketimpangan kelas ekstrem
sample_weights = compute_sample_weight(class_weight='balanced', y=y_train)
```

```python
# 2. Tingkatkan kedalaman pohon (max_depth=7) dan jumlah pohon (n_estimators=150)
num_classes = len(np.unique(y_train))
xgb_model = xgb.XGBClassifier(
    n_estimators=150,
    max_depth=7,               # Naik dari 5 ke 7 untuk memetakan batas keputusan rumit
    learning_rate=0.08,
    subsample=0.85,
    colsample_bytree=0.85,
    objective='multi:softprob',
    random_state=42,
    eval_metric='mlogloss',
    n_jobs=-1
)
```

```python
# 3. Latih model dengan memasukkan sample_weight
xgb_model.fit(X_train, y_train, sample_weight=sample_weights)
```

```python
# 4. Evaluasi ulang
y_pred = xgb_model.predict(X_test)
acc = accuracy_score(y_test, y_pred)
macro_f1 = f1_score(y_test, y_pred, average='macro')

print("\n" + "="*60)
print(f"Akurasi Baru Keseluruhan : {acc * 100:.2f}%")
print(f"Macro F1-Score Baru      : {macro_f1:.4f}")
print("="*60)

target_names = [str(cls) for cls in label_encoder.classes_]
print(classification_report(y_test, y_pred, target_names=target_names, digits=4))
```

```python
print("--- TAHAP XAI: EVALUASI TREESHAP VS LIME PADA MULTI-KELAS ---")

# ==============================================================================
# 1. PERSIAPAN SUBSET EVALUASI (150 SAMPEL TERSTRATIFIKASI)
# ==============================================================================
eval_size = 150
# Ambil sampel acak terstratifikasi dari data uji
_, X_eval, _, y_eval = train_test_split(
    X_test, y_test,
    test_size=eval_size,
    random_state=42,
    stratify=y_test
)
X_eval = X_eval.reset_index(drop=True)
feature_names = list(X_train.columns)
class_names = [str(cls) for cls in label_encoder.classes_]

print(f"Subset evaluasi siap: {len(X_eval)} sampel aliran paket.")
```

```python
# ==============================================================================
# 2. EKSTRAKSI TREESHAP & PENCATATAN LATENSI
# ==============================================================================
print("\nMengeksekusi TreeSHAP...")
tree_explainer = shap.TreeExplainer(xgb_model)

t0_shap = time.time()
shap_values_raw = tree_explainer.shap_values(X_eval)
t1_shap = time.time()

shap_latency_ms = ((t1_shap - t0_shap) / eval_size) * 1000
print(f"TreeSHAP selesai. Latensi rata-rata: {shap_latency_ms:.4f} ms per aliran.")
```

```python
# ==============================================================================
# 3. EKSTRAKSI LIME TABULAR & PENCATATAN LATENSI
# ==============================================================================
print("\nMengeksekusi LIME Tabular (membutuhkan waktu ~10-20 detik)...")
lime_explainer = lime_tabular.LimeTabularExplainer(
    training_data=np.array(X_train),
    feature_names=feature_names,
    class_names=class_names,
    mode='classification',
    kernel_width=np.sqrt(len(feature_names)) * 0.75,
    random_state=42
)

t0_lime = time.time()
lime_attributions = []

for i in range(eval_size):
    x_row = X_eval.iloc[i].values
    # Tentukan kelas yang diprediksi oleh model untuk instans ini
    pred_class = xgb_model.predict(x_row.reshape(1, -1))[0]

    # Ekstrak penjelasan LIME khusus untuk kelas prediksi tersebut
    exp = lime_explainer.explain_instance(
        x_row,
        xgb_model.predict_proba,
        labels=[pred_class],
        num_features=len(feature_names)
    )

    exp_dict = dict(exp.as_list(label=pred_class))
    weights = np.zeros(len(feature_names))

    # Petakan bobot LIME kembali ke urutan kolom fitur
    for f_idx, f_name in enumerate(feature_names):
        for cond_str, w in exp_dict.items():
            if f_name in cond_str:
                weights[f_idx] = w
                break
    lime_attributions.append(weights)

t1_lime = time.time()
lime_attributions = np.array(lime_attributions)
lime_latency_ms = ((t1_lime - t0_lime) / eval_size) * 1000
print(f"LIME selesai. Latensi rata-rata: {lime_latency_ms:.4f} ms per aliran.")
```

```python
# ==============================================================================
# 4. PERHITUNGAN DISAGREEMENT & FIDELITAS (COMPREHENSIVENESS & SUFFICIENCY)
# ==============================================================================
print("\nMenghitung metrik kuantitatif kualitas penjelasan...")
k_features = 5
fa_list, ra_list = [], []
comp_shap_list, suff_shap_list = [], []
comp_lime_list, suff_lime_list = [], []

# Baseline imputasi: nilai median dari data latih
baseline = X_train.median().values

for idx in range(eval_size):
    x_inst = X_eval.iloc[idx].values
    pred_probs = xgb_model.predict_proba(x_inst.reshape(1, -1))[0]
    pred_class = np.argmax(pred_probs)
    orig_prob = pred_probs[pred_class]

    # Ekstraksi bobot SHAP untuk kelas yang diprediksi
    if isinstance(shap_values_raw, list):
        s_weights = np.abs(shap_values_raw[pred_class][idx])
    elif shap_values_raw.ndim == 3:
        s_weights = np.abs(shap_values_raw[idx, :, pred_class])
    else:
        s_weights = np.abs(shap_values_raw[idx])

    l_weights = np.abs(lime_attributions[idx])

    # Indeks 5 fitur terpenting
    top_s = np.argsort(s_weights)[-k_features:]
    top_l = np.argsort(l_weights)[-k_features:]

    # A. Feature Agreement (FA)
    overlap = set(top_s).intersection(set(top_l))
    fa = len(overlap) / float(k_features)
    fa_list.append(fa)

    # B. Rank Agreement (RA) - Kendall Tau
    common = list(overlap)
    if len(common) > 1:
        rank_s = [np.where(top_s == f)[0][0] for f in common]
        rank_l = [np.where(top_l == f)[0][0] for f in common]
        tau, _ = stats.kendalltau(rank_s, rank_l)
        ra_list.append(0.0 if np.isnan(tau) else tau)
    else:
        ra_list.append(0.0)

    # C. Metrik Fidelitas TreeSHAP
    # Comprehensiveness: Hapus top fitur -> amati penurunan probabilitas
    x_comp_s = x_inst.copy()
    x_comp_s[top_s] = baseline[top_s]
    prob_after_comp_s = xgb_model.predict_proba(x_comp_s.reshape(1, -1))[0, pred_class]
    comp_shap_list.append(orig_prob - prob_after_comp_s)

    # Sufficiency: Sisakan HANYA top fitur -> amati deviasi probabilitas
    x_suff_s = baseline.copy()
    x_suff_s[top_s] = x_inst[top_s]
    prob_after_suff_s = xgb_model.predict_proba(x_suff_s.reshape(1, -1))[0, pred_class]
    suff_shap_list.append(orig_prob - prob_after_suff_s)

    # D. Metrik Fidelitas LIME
    x_comp_l = x_inst.copy()
    x_comp_l[top_l] = baseline[top_l]
    prob_after_comp_l = xgb_model.predict_proba(x_comp_l.reshape(1, -1))[0, pred_class]
    comp_lime_list.append(orig_prob - prob_after_comp_l)

    x_suff_l = baseline.copy()
    x_suff_l[top_l] = x_inst[top_l]
    prob_after_suff_l = xgb_model.predict_proba(x_suff_l.reshape(1, -1))[0, pred_class]
    suff_lime_list.append(orig_prob - prob_after_suff_l)
```

```python
# ==============================================================================
# 5. TABULASI HASIL UNTUK PAPER KONFERENSI
# ==============================================================================
print("\n" + "="*65)
print("HASIL EVALUASI FORMAL XAI PADA KLASIFIKASI TRAFIK JARINGAN")
print("="*65)
print(f"Rata-rata Feature Agreement (Top-{k_features}) : {np.mean(fa_list):.4f} (+/- {np.std(fa_list):.4f})")
print(f"Rata-rata Rank Agreement (Kendall-Tau)  : {np.mean(ra_list):.4f} (+/- {np.std(ra_list):.4f})")
print("-"*65)
print(f"TreeSHAP Comprehensiveness (Tinggi=Baik) : {np.mean(comp_shap_list):.4f}")
print(f"LIME Comprehensiveness     (Tinggi=Baik) : {np.mean(comp_lime_list):.4f}")
print("-"*65)
print(f"TreeSHAP Sufficiency       (Rendah=Baik) : {np.mean(suff_shap_list):.4f}")
print(f"LIME Sufficiency           (Rendah=Baik) : {np.mean(suff_lime_list):.4f}")
print("-"*65)
print(f"Latensi Inferensi TreeSHAP per Aliran    : {shap_latency_ms:.4f} ms")
print(f"Latensi Inferensi LIME per Aliran        : {lime_latency_ms:.4f} ms")
print("="*65)
```
